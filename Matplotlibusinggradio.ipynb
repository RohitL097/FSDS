```python
import pandas as pd
```

```python
data = {
    "Month":["Jan","Feb","March","April","May","June","July","August","September","October","November","December"],
    "Sales":[10000,20000,15000,16000,20000,13000,9000,28000,21000,5500,13500,14000],
    "Profit":[2000,3200,1900,4000,2300,5000,7800,10000,15050,1600,958,10000]
}
```

```python
len(data["Sales"])
len(data["Profit"])
```

```python
len(data["Month"])
```

```python
data
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
df.head(5)
```

```python
df.tail(3)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize = (8,6))
plt.plot(df["Month"],df["Sales"],color = "blue",marker = 'o',linestyle = '-',label = 'Sales')
plt.title('Sales trends over months')
plt.xlabel('Month')
plt.ylabel('Sales ($)')
plt.grid(True)
plt.legend()

#Line plot monthly sales
```

```python
plt.figure(figsize = (8,5))
plt.plot(df['Month'],df['Profit'],color = 'blue',marker = 'o',linestyle = '-',label = 'Profit')
plt.xlabel('Month')
plt.ylabel('Profit')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.bar(df['Month'],df['Sales'],color = "orange")
```

```python
plt.bar(df['Month'],df['Profit'],color = 'blue')
```

```python
plt.figure(figsize = (8,5))
plt.pie(df['Profit'],labels = df['Month'],autopct = '%1.1f%%', startangle = 140, colors = plt.cm.Paired.colors)
plt.title('Profit distribution by month')
plt.tight_layout()
plt.show()
```

```python
plt.hist(df['Month'], bins = 5, color = 'purple',edgecolor = 'Black')
plt.title('A histogram of sales over months')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.tight_layout()
#here is how we create a histogram with data
```

```python
plt.figure(figsize=(10, 6))
width = 0.3
plt.bar(df['Month'], df['Sales'], width=width, label='Sales', color='skyblue')
plt.bar(df['Month'], df['Profit'], width=width, label='Profit', color='orange', bottom=df['Sales'])
plt.title('Sales and Profit Comparison by Month')
plt.xlabel('Month')
plt.ylabel('Amount ($)')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['Sales'], df['Profit'], color='green', s=100, edgecolors='black')
plt.title('Sales vs Profit')
plt.xlabel('Sales ($)')
plt.ylabel('Profit ($)')
plt.grid(True)
plt.tight_layout()
plt.show()
```
