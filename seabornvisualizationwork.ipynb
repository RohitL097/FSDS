```python
sns.lineplot(data = tips, x = 'size' , y = 'total_bill', hue = 'sex', ci = None, markers = 'o')
```

```python
sns.barplot(data = tips,x = 'day', y = 'total_bill', hue = 'sex', palette = 'muted')
plt.title('Bar plot of day wise bills by gender')
plt.show()
```

```python
sns.boxplot(data = tips, x = 'day', y = 'tip', hue = 'sex', palette = 'Set2')
plt.title('Day wise tips by gender box plot')
plt.show()
```

```python
sns.countplot(data = tips, x = 'sex', hue = 'tip',color = 'orange')
```

```python
sns.countplot(data = tips, x = 'sex', hue = 'smoker', palette = 'dark')
```
